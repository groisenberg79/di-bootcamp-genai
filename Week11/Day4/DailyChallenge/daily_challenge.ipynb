1. Understand Classification Types

- Read about Binary Classification, Multi-class Classification, and Multi-label Classification.
- Write a brief explanation of each type with an example.

#### Solution

- Binary Classification:
    -   Definition: The simplest form, classifying instances into exactly one of two classes.
    -   Examples: Spam detection (Spam/Not Spam), Cancer diagnosis (Benign/Malignant).
    -   Output: A single probability score or a binary value (0 or 1).
    -   Activation: Sigmoid.

- Multi-class Classification:
    -   Definition: Classifying instances into exactly one class out of three or more categories.
    -   Examples: Digit recognition (0-9), Language identification, Fruit classification (Apple/Banana/Orange).
    -   Output: Probabilities that sum to 1 across all classes.
    -   Activation: Softmax.

- Multi-label Classification:
    -   Definition: Assigns a sample to multiple non-mutually exclusive labels simultaneously. An item can have no labels, one, or several.
    -   Examples: Document classification (a news article being both "politics" and "finance"), Image tagging ("sky", "ocean", "sunny").
    -   Output: A binary vector indicating the presence/absence of each label (e.g., [1, 0, 1, 1]).
    -   Activation: Independent Sigmoid for each label.

2. Set Up Your Python Environment and dataset

- Ensure you have the required libraries installed:

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf
from sklearn.datasets import make_circles
from sklearn.model_selection import train_test_split
from keras.models import Sequential
```

It is time for creating a dataset to work on:

```python
samples = 1000
X, y = make_circles(samples,
                    noise = 0.03,
                    random_state = 42)

print('X : ', X)
print('\n')
print('y : ', y)
```

Visualize the dataset using scatter plots to understand the data distribution

#### Solution

```python
plt.scatter(X[:, 0], X[:, 1], c=y, cmap='coolwarm')
plt.show()
```

3. Build a Basic Neural Network Model

- Create a Sequential Model in TensorFlow with one dense layer.
- Use Binary Crossentropy as the loss function.
- Train the model and check the accuracy.

#### Solution

```python
# Build model structure
model_1 = Sequential([
    tf.keras.Input(shape=(2,)),
    tf.keras.layers.Dense(1, activation='sigmoid')
])

# Configure the model for training
model_1.compile(optimizer='adam',
              loss='binary_crossentropy',
              metrics=['accuracy'])

# 
model_1.fit(X, y, epochs=20)

model_1.evaluate(X, y)
```

4. Improve the Model

- Add more layers and neurons to the model.
- Train the model for more epochs.
- Try different optimizers like Adam instead of SGD.

```python
# Build model structure
model_2 = Sequential([
    tf.keras.Input(shape=(2,)),
    tf.keras.layers.Dense(units=256, activation='relu'),
    tf.keras.layers.Dense(1, activation='sigmoid')
])

# Configure the model for training
model_2.compile(optimizer='adam',
              loss='binary_crossentropy',
              metrics=['accuracy'])

# 
model_2.fit(X, y, epochs=25)

model_2.evaluate(X, y)
```

```python
# Build model structure
model_3 = Sequential([
    tf.keras.Input(shape=(2,)),
    tf.keras.layers.Dense(units=256, activation='relu'),
    tf.keras.layers.Dense(units=128, activation='relu'),
    tf.keras.layers.Dense(1, activation='sigmoid')
])

# Configure the model for training
model_3.compile(optimizer='adam',
              loss='binary_crossentropy',
              metrics=['accuracy'])

# 
model_3.fit(X, y, epochs=25)

model_3.evaluate(X, y)
```

5. Visualize the Decision Boundary

- Implement a function plot_decision_boundary() to visualize model predictions.
- Plot decision boundaries for different models and compare results.

#### Solution

```python
def plot_decision_boundary(model, X, y):
    # Define the limits of the plot
    x_min, x_max = X[:, 0].min() - 0.1, X[:, 0].max() + 0.1
    y_min, y_max = X[:, 1].min() - 0.1, X[:, 1].max() + 0.1

    # Create a grid of points covering the whole area
    xx, yy = np.meshgrid(
        np.linspace(x_min, x_max, 300),
        np.linspace(y_min, y_max, 300)
    )

    # Stack the grid points into shape (num_points, 2)
    grid = np.c_[xx.ravel(), yy.ravel()]

    # Get model predictions for every point in the grid
    y_pred = model.predict(grid, verbose=0)

    # Convert probabilities to class labels 0 or 1
    y_pred = (y_pred > 0.5).astype(int)

    # Reshape predictions back to the grid shape
    y_pred = y_pred.reshape(xx.shape)

    # Plot the decision regions
    plt.contourf(xx, yy, y_pred, alpha=0.5, cmap=plt.cm.RdBu)

    # Plot the original data points
    plt.scatter(X[:, 0], X[:, 1], c=y, s=40, cmap=plt.cm.RdBu, edgecolors='k')

    plt.xlim(xx.min(), xx.max())
    plt.ylim(yy.min(), yy.max())
```

```python
plt.figure(figsize=(6, 5))
plot_decision_boundary(model_1, X, y)
plt.title("Decision Boundary - Model 1")
plt.show()

plt.figure(figsize=(6, 5))
plot_decision_boundary(model_2, X, y)
plt.title("Decision Boundary - Model 2")
plt.show()

plt.figure(figsize=(6, 5))
plot_decision_boundary(model_3, X, y)
plt.title("Decision Boundary - Model 3")
plt.show()
```

**Observation**: model_1 learns only a linear decision boundary, so it cannot properly separate the two concentric circles.
model_2, with one hidden ReLU layer, learns a nonlinear boundary that fits the circular structure much better.
model_3, with two hidden layers, also learns a nonlinear boundary and achieves an even better fit, showing that deeper networks can model more complex patterns.
