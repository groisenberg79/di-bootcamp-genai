## 🌟 Exercise 5 : Indexing Arrays
### Instructions
Select the second row from a given 2D NumPy array.

```python
arr = np.arange(1,10).reshape(3,3)
print("Array:")
print(arr)
print()
print("Second row:")
print(arr[1, :])
```

## 🌟 Exercise 6 : Reversing elements
### Instructions
Reverse the order of elements in a given 1D NumPy array (first element becomes last).

```python
arr = np.arange(10)
print("Original array:")
print(arr)
print()

new_arr = arr[::-1]
print("New Array:")
print(new_arr)
```

## 🌟 Exercise 8 : Simple Aggregate Funcs
### Instructions
Find the sum and average of a given 1D array.

```python
arr = np.arange(10)
print("Array:")
print(arr)
print()

print("Sum of elements:")
print(arr.sum())
print

print("Average of elements:")
print(arr.mean())
```

## 🌟 Exercise 10 : Conditional Selection of Values
### Instructions
Extract all odd numbers from a given NumPy array.

```python
arr = np.arange(10)
print("Array:")
print(arr)
print("Odd numbers:")
print(arr[arr % 2 == 1])
```
