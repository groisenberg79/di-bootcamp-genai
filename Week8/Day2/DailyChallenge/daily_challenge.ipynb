```python
# Import modules
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import calendar
```

```python
# Generate the numpy ndarray with random data
np.random.seed(42)
monthly_temps = np.random.uniform(-5, 35, (10,12)).round()
print(monthly_temps)
```

```python
# Create the dataframe
cities = ['City 1', 'City 2', 'City 3', 'City 4', 'City 5', 'City 6', 'City 7', 'City 8', 'City 9', 'City 10']

month_names = list(calendar.month_abbr)[1:]

df = pd.DataFrame(data=monthly_temps, index=cities, columns=month_names)
print(df)
```

```python
# Check mean temperature for each month
print("Mean temperature in each month (Celsius):")
print(df.mean(axis=0))
```

```python
# Check mean temperature for each city
print("Mean temperatures per city (Celsius):")
mean_per_city = df.mean(axis=1)
print(mean_per_city)
print()
# Find city with lowest and highest mean temperatures
print(f"City with highest mean temperature: {mean_per_city.idxmax()}")
print(f"City with lowest mean temperature: {mean_per_city.idxmin()}")
print()
```

```python
# Plot graph with temperature variation for each city along the year
plt.figure(figsize=(10,6))

# Create a line for each city
for city in df.index:
    plt.plot(df.columns, df.loc[city], marker="o", label=city)

plt.legend(bbox_to_anchor=(1.05,1))
plt.xticks(rotation=45)
plt.title("Monthly Temperature by City", fontsize=16, fontweight='bold')
plt.xlabel("Months")
plt.ylabel("Temperature (Celsius)")
plt.tight_layout()
plt.show()
```

**Observation**: Since the dataset was generated randomly, there are no meaningful seasonal trends or consistent patterns observable in the plot.
