### Solution

#### Problem Statement

- The objective of this project is to predict whether a loan applicant will default on a loan (target variable: loan default status, binary: default / no default) in order to support lending decisions such as loan approval, interest rate assignment, and credit limit determination.

- Accurate prediction of loan default risk enables financial institutions to reduce financial losses, price risk appropriately, and maintain regulatory compliance while continuing to extend credit responsibly.

#### Data Types Required

To build a predictive model for loan default, several categories of data would be required.

**Personal and Demographic Information**

- This includes variables such as age, employment status, occupation, marital status, education level, and residential stability. These factors can help assess income stability and financial responsibility, which may correlate with repayment ability.

**Financial and Income Information**

- This includes monthly income, total assets, outstanding debts, debt-to-income ratio, and savings balances. These variables directly measure the borrower’s financial capacity to repay the loan.

**Credit History Information**

- This includes credit score, number of previous loans, repayment history, number of missed payments, credit utilization ratio, and length of credit history. Past repayment behavior is one of the strongest predictors of future default risk.

**Loan-Specific Information**

- This includes loan amount, interest rate, loan term, loan purpose, and collateral (if applicable). Loan structure influences repayment burden, and certain loan purposes may carry different levels of risk.

**Behavioral and Transactional Data**

- This may include account transaction patterns, spending behavior, overdraft frequency, and recent changes in income or expenses. Behavioral indicators can provide early signals of financial distress that are not visible in static credit reports.

#### Data Sources and Collection Methods

**Internal Financial Institution Records**

- The primary source of data would be the lending institution’s internal databases, which contain applicant information, loan contracts, repayment history, and account activity.
These data can be extracted from operational databases and integrated into a centralized analytics environment after appropriate anonymization.

**Credit Bureau Data**

- Credit bureaus provide standardized credit scores, repayment history across institutions, and credit exposure information. This data can be integrated via secure APIs or periodic data feeds, subject to regulatory and contractual agreements.

**Customer Application Forms**

- Loan application forms provide self-reported income, employment, and personal information.
These data are collected at the time of application and stored in structured digital systems for further processing.

**External Economic Data**

- Macroeconomic indicators such as unemployment rates, inflation, or regional economic conditions may also be incorporated. These data can be obtained from public statistical agencies and linked to borrowers by geographic region and time period.

#### Risks and Constraints

Loan default prediction involves significant regulatory and ethical considerations. Privacy regulations such as data protection laws require secure handling, anonymization, and limited use of personal data. There is also a risk of sampling bias if the dataset overrepresents certain demographic groups or historical lending practices that may reflect discriminatory patterns. Data quality issues, such as missing values or inaccurate self-reported income, may reduce model reliability. Governance processes must ensure that the model is transparent, auditable, and compliant with financial regulations, particularly if it influences approval decisions or interest rate setting.

```python

# This piece of code is already prefilled, run it to execute it and see the results.
# It demonstrates standard classification metrics for binary loan default prediction.

from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, average_precision_score, confusion_matrix, classification_report

# Please replace these placeholders with your true labels and predicted probabilities.
y_true = [0,1,0,1,0,0,1,0,1,0]            # placeholder labels
y_pred_proba = [0.05,0.80,0.10,0.65,0.20,0.15,0.70,0.30,0.85,0.25]  # placeholder probabilities

# You should set a decision threshold that reflects the precision–recall trade-off for your business case.
threshold = 0.5
y_pred = [1 if p >= threshold else 0 for p in y_pred_proba]

print("Accuracy:", round(accuracy_score(y_true, y_pred), 4))
print("Precision:", round(precision_score(y_true, y_pred, zero_division=0), 4))
print("Recall:", round(recall_score(y_true, y_pred, zero_division=0), 4))
print("F1-score:", round(f1_score(y_true, y_pred, zero_division=0), 4))
print("ROC-AUC:", round(roc_auc_score(y_true, y_pred_proba), 4))
print("PR-AUC (Average Precision):", round(average_precision_score(y_true, y_pred_proba), 4))
print("\nConfusion matrix:\n", confusion_matrix(y_true, y_pred))
print("\nClassification report:\n", classification_report(y_true, y_pred, zero_division=0))
```

### Solution

1. **Candidate models**:
A good baseline model for loan default prediction is Logistic Regression because it is fast, interpretable, and produces calibrated probabilities that are useful for risk decisions. A strong second model is a tree-based model such as Random Forest or Gradient Boosting because it can capture non-linear relationships and interactions among features that are common in credit risk data.

2. **Evaluation plan**:
I would split the dataset into training and test sets (for example, 80% train and 20% test) using a stratified split so the proportion of defaults is similar in both sets. On the training set, I would use stratified k-fold cross-validation (e.g., 5 folds) to estimate performance robustly and reduce dependence on a single split. I would report metrics that reflect both ranking quality and error costs, including ROC-AUC, PR-AUC (Average Precision), precision, recall, and F1-score, and I would also inspect the confusion matrix. I would choose a decision threshold by comparing precision–recall trade-offs and selecting the cutoff that matches the business cost of false negatives (missed defaults) versus false positives (rejecting good borrowers).

3. **Class imbalance handling**:
Because loan defaults are typically less frequent than non-defaults, I would use stratification in both the train/test split and cross-validation to preserve the default rate. I would also try class weights (e.g., “balanced” weights) so the model penalizes misclassifying defaults more heavily than misclassifying non-defaults. If needed, I would additionally test resampling strategies (such as oversampling the default class) applied only to the training folds to avoid biasing evaluation.

4. **Hyperparameter iteration and avoiding leakage**:
I would tune hyperparameters using cross-validation on the training data only (for example with grid search or randomized search) and keep the test set untouched until the final evaluation. To avoid data leakage, I would ensure preprocessing steps such as imputation, encoding, scaling, and resampling are performed inside a training-only pipeline so that no information from the test set influences training.
