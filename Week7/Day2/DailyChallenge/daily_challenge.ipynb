## 2. Data Visualization with Matplotlib:
- Create an interactive line chart to show sales trends over the years.
- Build an interactive map to visualize sales distribution by country.

### Solution

**Observation**: Since the exercise asks for an interactive chart, I decided to use plotly instead of matplotlib.

```python
# Group by year -> Select Sales -> Sum sales for each year -> Create DataFrame
grouped = df.groupby(df['Ship Date'].dt.year)['Sales'].sum().reset_index()

# Plot line chart 
fig = px.line(
    grouped, 
    x='Ship Date', 
    y='Sales', 
    title='Yearly Sales Trend',
    markers=True
)
fig.update_layout(xaxis_title='Year', 
                  yaxis_title='Total Sales ($)')
fig.show()
```

```python
df["Year"] = df["Ship Date"].dt.year
sales_per_year = (
    df.groupby("Year")
      .size()
      .reset_index(name="Sales Count")
)
print(sales_per_year)
```

**Observation**: As shown above, the sales count for 2018 is very small compared to the previous years. That is possibly due to the small sample for 2018, and not due to an actual drop in sales.

```python
# Aggregate sales by country
country_sales = (
    df.groupby("Country", as_index=False)["Sales"]
      .sum()
      .rename(columns={"Sales": "TotalSales"})
)

# Choropleth map (Plotly will try to match country names)
fig = px.choropleth(
    country_sales,
    locations="Country",
    locationmode="country names",
    color="TotalSales",
    hover_name="Country",
    color_continuous_scale="Blues",
    title="Total Sales by Country"
)

fig.update_layout(margin=dict(l=0, r=0, t=50, b=0))
fig.show()
```

**Observation**: The dataset contains only United States sales records, so the country-level interactive map displays a single-country distribution. A state-level map would provide more meaningful geographic insight.

## 3. Data Visualization with Seaborn:
- Use Seaborn to generate a bar chart showing top 10 products by sales.
- Create a scatter plot to analyze the relationship between profit and discount.

### Solution

```python
product_grouped = df.groupby("Product Name")["Sales"].sum().reset_index().sort_values(by='Sales', ascending=False).head(10)

fig, ax = plt.subplots(figsize=(10,6))
ax = sns.barplot(data=product_grouped, x='Product Name', y='Sales')
ax.tick_params(axis='x', rotation=90)

plt.show()
```

```python
# Plot scatterplot
sns.lmplot(x='Discount', y='Profit', data=df, line_kws={"color": "red"})
plt.show()
```

```python
# Check correlation between Discounts and Profits
df[["Discount","Profit"]].corr()
```

**Observation**: The scatter plot shows a weak negative relationship between discount and profit. The correlation coefficient (r ≈ −0.22) indicates that higher discounts are associated with lower profits, but the effect size is small and profit variability is high, which explains the nearly flat regression line.
